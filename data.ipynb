```python
# Import thư viện pandas để xử lý dữ liệu
import pandas as pd


# ==============================
# BƯỚC 1: ĐỌC 2 DATASET CSV
# ==============================

# Đọc dataset thứ nhất
# Dataset này chứa dữ liệu lương ngành AI/Data với số lượng lớn
df1 = pd.read_csv(r'D:\FPT\Do_An_Tot_Nghiep\DATA\salaries.csv')

# Đọc dataset thứ hai
# Dataset này là dữ liệu bổ sung để tăng độ đa dạng dữ liệu
df2 = pd.read_csv(r'D:\FPT\Do_An_Tot_Nghiep\DATA\ds_salaries.csv')

# Đọc dataset thứ ba
df3 = pd.read_json(r'D:\FPT\Do_An_Tot_Nghiep\Data Science, AI & ML Job Salaries in 2025\salaries.json')
```

```python

# ==============================
# BƯỚC 2: KIỂM TRA KÍCH THƯỚC DỮ LIỆU
# ==============================

# Tạo bảng thống kê kích thước dữ liệu
shape_df = pd.DataFrame({
    'Dataset': ['Dataset 1', 'Dataset 2', 'Dataset 3'],
    'Rows': [df1.shape[0], df2.shape[0], df3.shape[0]],
    'Columns': [df1.shape[1], df2.shape[1], df3.shape[1]] 
})

# Hiển thị bảng
shape_df
```

```python
# ==============================
# BƯỚC 3: KIỂM TRA TÊN CỘT
# ==============================

# Hiển thị danh sách tên cột của dataset 1
print(df1.columns)

# Hiển thị danh sách tên cột của dataset 2
print(df2.columns)

# Hiển thị danh sách tên cột của dataset 3
print(df3.columns)
```

```python
# ==============================
# BƯỚC 4: GỘP 2 DATASET
# ==============================

# Sử dụng hàm concat() để nối 2 bảng dữ liệu lại với nhau
# ignore_index=True giúp đánh lại số thứ tự index mới
merged_df = pd.concat([df1, df2, df3], ignore_index=True)
```

```python
# ==============================
# BƯỚC 5: KIỂM TRA SAU KHI MERGE
# ==============================

# Hiển thị kích thước dữ liệu sau khi merge
print("Merged dataset shape:", merged_df.shape)

# Xóa cột 'Unnamed: 0' (cột rác sinh ra từ file cũ) nếu nó tồn tại
if 'Unnamed: 0' in merged_df.columns:
    merged_df = merged_df.drop(columns=['Unnamed: 0'])

# Hiển thị 5 dòng đầu tiên của dữ liệu
print(merged_df.head())
```

```python
# ==============================
# BƯỚC 6: XÓA DỮ LIỆU TRÙNG LẶP
# ==============================

# Xóa các dòng dữ liệu bị trùng
merged_df.drop_duplicates(inplace=True)

# Kiểm tra lại kích thước sau khi xóa dữ liệu trùng
print("After removing duplicates:", merged_df.shape)
```

```python
# ==============================
# BƯỚC 7: LƯU DATASET MỚI
# ==============================

# Lưu dataset đã merge thành file CSV mới
merged_df.to_csv(
    r'D:\FPT\Do_An_Tot_Nghiep\DATA\merged_ai_salary_dataset.csv',
    index=False
)

# Thông báo hoàn thành
print("Merged dataset saved successfully!")
```
